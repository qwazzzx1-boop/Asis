# 📱 Сборка Voice Assistant APK в Google Colab

### Исправление ошибки `kivy==2.3.0`:
В python-for-android нельзя указывать версию у `kivy` (например `kivy==2.3.0`), иначе сборщик не находит внутренний рецепт (recipe) и пытается собрать Kivy через сырой pip, что вызывает `Error code: 1`.

В файле `buildozer.spec` зависимость исправлена на чистый `requirements = python3,kivy,openssl`.

### Запуск:
Запустите все ячейки по очереди или нажмите **Среда выполнения -> Выполнить всё**.

```python
# 1. Системные пакеты, компиляторы и совместимый Cython 0.29.33
!sudo apt-get update -qq
!sudo apt-get install -y -qq git zip unzip openjdk-17-jdk autoconf libtool pkg-config zlib1g-dev libncurses5-dev libncursesw5-dev libtinfo5 cmake libffi-dev libssl-dev libltdl-dev --fix-missing
!pip install --upgrade "Cython==0.29.33" virtualenv
!pip install --upgrade git+https://github.com/kivy/buildozer.git
```

```python
# 2. Загрузка и распаковка проекта
import os, zipfile, glob
from google.colab import files

%cd /content
if not os.path.exists('voice-assistant-android.zip'):
    print('Пожалуйста, выберите файл voice-assistant-android.zip:')
    files.upload()

!rm -rf /content/voice-assistant-android
with zipfile.ZipFile('voice-assistant-android.zip', 'r') as zip_ref:
    zip_ref.extractall('/content/voice-assistant-android')

%cd /content/voice-assistant-android
!ls -la
```

```python
# 3. Удаление старого повреждённого кэша и чистая сборка APK
!rm -rf .buildozer
!buildozer -v android debug
```

```python
# 4. Автоматическое скачивание собранного APK
import glob
from google.colab import files

apk_files = glob.glob('/content/voice-assistant-android/bin/*.apk')
if apk_files:
    print(f'✅ Успех! Файл готов: {apk_files[0]}')
    files.download(apk_files[0])
else:
    print('❌ APK файл не найден в папке bin/!')
```
